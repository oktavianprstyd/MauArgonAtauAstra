## 1. Business Understanding

### 1.1 Latar Belakang Masalah
Dalam era transformasi digital, toko daring (*e-commerce*) menghadapi tantangan besar terkait tingkat konversi pengunjung. Banyak pengguna menjelajahi berbagai halaman produk namun keluar tanpa melakukan transaksi (*cart abandonment*). Mengidentifikasi sinyal perilaku navigasi pengunjung selama sesi belanja daring menjadi krusial agar pelaku bisnis dapat melakukan intervensi pemasaran yang tepat sasaran.

### 1.2 Rumusan Masalah
Bagaimana memprediksi niat pembelian pengunjung toko online (apakah sesi menghasilkan transaksi `Revenue = True` atau tidak `Revenue = False`) secara akurat berdasarkan pola navigasi web dan metrik perilaku sesi?

### 1.3 Pertanyaan Bisnis (Business Questions)
1. Fitur aktivitas navigasi apa (misalnya durasi melihat produk, Bounce Rates, Exit Rates, dan Page Values) yang memiliki korelasi tertinggi terhadap transaksi pembelian?
2. Bagaimana dampak faktor temporal seperti bulan kunjungan, akhir pekan, atau kedekatan dengan hari khusus (*Special Day*) terhadap kecenderungan berbelanja?
3. Model klasifikasi berbasis materi praktikum (Support Vector Machine / SVM, Logistic Regression, KNN) manakah yang memberikan performa terbaik dalam mengatasi ketidakseimbangan kelas (*class imbalance*)?

### 1.4 Tujuan Proyek
- Menganalisis pola perilaku sesi melalui eksplorasi data komprehensif (EDA).
- Membangun model machine learning klasifikasi biner untuk memprediksi niat beli pengunjung toko online.
- Menyajikan wawasan bisnis operasional bagi toko daring guna mengoptimalkan konversi penjualan.

### 1.5 Pendekatan & Metrik Evaluasi
- **Metode**: Supervised Learning — Binary Classification.
- **Metrik Evaluasi**: Mengingat target kelas tidak seimbang (~84.5% False vs ~15.5% True), evaluasi difokuskan pada **F1-Score (Macro / Class True)**, **Precision**, **Recall**, serta kurva **ROC-AUC** dan **Confusion Matrix**.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Konfigurasi gaya visualisasi
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)

# Membaca dataset dari folder data
data_path = "../data/online_shoppers_intention.csv"
df = pd.read_csv(data_path)
print(f"Dimensi Dataset: {df.shape[0]} baris, {df.shape[1]} kolom")
df.head()
```

## 3. Exploratory Data Analysis (EDA)

Tahap ini mencakup analisis statistik deskriptif dan visualisasi untuk menemukan pola-pola kunci pada perilaku pengunjung web.

```python
# Distribusi Target Label: Revenue
fig, ax = plt.subplots(1, 2, figsize=(12, 5))
df["Revenue"].value_counts().plot(kind="bar", ax=ax[0], color=["#3498db", "#e74c3c"])
ax[0].set_title("Distribusi Frekuensi Revenue")
ax[0].set_xlabel("Revenue (True / False)")
ax[0].set_ylabel("Jumlah Sesi")

df["Revenue"].value_counts().plot(kind="pie", autopct="%1.2f%%", ax=ax[1], colors=["#3498db", "#e74c3c"], startangle=90)
ax[1].set_title("Proporsi Target Revenue")
ax[1].set_ylabel("")
plt.tight_layout()
plt.show()

print("Persentase Kelas:")
print(df["Revenue"].value_counts(normalize=True) * 100)
```
